```python
import kagglehub
import pandas as pd
import os

# Download dataset
path = kagglehub.dataset_download("datatattle/covid-19-nlp-text-classification")

# See files
print(os.listdir(path))
```

```python
df = pd.read_csv(os.path.join(path, "Corona_NLP_test.csv"), encoding='latin1')
df.head()
```

```python
from nltk.corpus import stopwords
text = df['OriginalTweet']
text
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
stemmer = PorterStemmer()
lemm = WordNetLemmatizer()
```

```python
import re
corpus = []
for i in range(len(text)):
    review = re.sub('[^a-zA-Z]',' ', text[i])
    review=review.lower()
    corpus.append(review)
```

```python
corpus
```

## Stemming

```python
for i in corpus:
    words=nltk.word_tokenize(i)
    for word in words:
        if word not in set(stopwords.words('english')):
            print(stemmer.stem(word))
```

## Lemmatization

```python
text = df['OriginalTweet']
```

```python
import re
corpus = []
for i in range(len(text)):
    review = re.sub('[^a-zA-Z]',' ', text[i])
    review= review.lower()
    review = review.split()
    review = [lemm.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review= ' '.join(review)
    corpus.append(review)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(binary = True, ngram_range = (2,3))
X = cv.fit_transform(corpus)
```

## TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
cv = TfidfVectorizer(ngram_range = (3,3) , max_features =10)
X = cv.fit_transform(corpus)
```
