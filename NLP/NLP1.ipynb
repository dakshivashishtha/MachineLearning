```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
## Tokenization -- converts para into sentences and then into words
nltk.download('stopwords')
nltk.download('punkt')
sent = nltk.sent_tokenize(paragraph)
```

```python
sent
```

```python
type(sent)
```

```python
stemmer = PorterStemmer()
```

```python
stemmer.stem('going')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize("going")
```

```python
lemmatizer.lemmatize("goes")
```

```python
len(sent)
```

```python
import re
corpus = []
for i in range(len(sent)):
    review=re.sub('[^a-zA-Z]',' ' , sent[i])
    review=review.lower()
    corpus.append(review)
```

```python
corpus
```

```python
## stemming
for i in corpus:
    words= nltk.word_tokenize(i)
    for word in words:
        if word not in set(stopwords.words('english')):
            print(stemmer.stem(word))
```

## For Bag of Words ,  we use Count Vectorizer


```python
## Apply Stopwords. Lemmatize
import re
corpus=[]
for i in range(len(sent)):
    review = re.sub('[^a-zA-Z]', ' ',sent[i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(binary =True, ngram_range =(2,3))
X = cv.fit_transform(corpus)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
cv = TfidfVectorizer(ngram_range= (3,3),max_features=10) # max_features=10, means keep only 10 most imp 3-grams from the corpus
X= cv.fit_transform(corpus)
```
